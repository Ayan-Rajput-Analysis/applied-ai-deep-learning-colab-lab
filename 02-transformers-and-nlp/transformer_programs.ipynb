```python
# Install required library


from transformers import pipeline

# Load text generation pipeline
generator = pipeline("text-generation", model="gpt2")

# Generate text
prompt = "Artificial Intelligence is transforming the world by"

result = generator(
    prompt,
    max_length=100,
    num_return_sequences=2,
    temperature=0.7,
    do_sample=True
)

# Display results
for i, output in enumerate(result):
    print(f"\n--- Generation {i+1} ---")
    print(output["generated_text"])
```

```python
from diffusers import StableDiffusionPipeline
import torch
from IPython.display import display

model = "runwayml/stable-diffusion-v1-5"

pipe = StableDiffusionPipeline.from_pretrained(
    model,
    torch_dtype=torch.float16
)

pipe = pipe.to("cuda")

prompt = "A high-action, cinematic wide shot of a young teenage boy playing cricket in a sun-drenched professional stadium. The boy is in the middle of a powerful cover drive stroke, wearing a classic white cricket uniform with subtle signs of grass stains. Sunlight is catching the dust particles in the air, creating a dramatic lens flare. High-speed photography style, sharp focus on the boy’s intense expression, blurred stadium crowd in the background, professional sports photography, 8k resolution, photorealistic, vibrant colors."

image = pipe(
    prompt,
    height = 600,
    width = 600,
    num_inference_steps=50,
    guidance_scale=7.5
).images[0]

display(image)
image.save('generated_image.png')
```
