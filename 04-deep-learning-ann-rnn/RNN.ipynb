```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, Embedding, SimpleRNN, Dense
```

```python
sentences = [
    "I love this product",
    "This movie made me smile",
    "Service was friendly and quick",
    "Today felt bright and happy",
    "This is the best day",
    "Absolutely fantastic experience",
    "I enjoyed every single moment",
    "Great job, well done",
    "The food tasted delicious",
    "Totally recommend to everyone",
    "Very satisfied with results",
    "This worked better than expected",
    "Amazing quality and value",
    "Such a pleasant surprise",
    "I feel positive about this",
    "I hate this product",
    'This movie bored me',
    'Service was rude and slow',
    'Today was cooled and lonely',
    'This is the worst day',
    'Terrible experince overall',
    'I regret buying this',
    'Very disappointing with results',
    'The food tasted awful',
    'Do not recommend this',
    'It broke after one use',
    'Not worth the money',
    'Utterly frusting and annoying',
    'I feel negative about this',
    'Such a waste of time',
]

labels = [1]*15 + [0]*15
labels = np.array(labels)
```

```python
vocab_size = 2000
tok = Tokenizer(num_words=vocab_size,oov_token='<OOV>')
tok.fit_on_texts(sentences)
seqe = tok.texts_to_sequences(sentences)
max_len = max((len(s)) for s in seqe)
X = pad_sequences(seqe, maxlen=max_len,padding='post')
y = labels
```

```python
embed_din = 16
rnn_units = 8

```

```python
inp = Input(shape = (max_len,),dtype='int32',name='input')
x = Embedding(input_dim=vocab_size,output_dim=embed_din,mask_zero=True,name='embed')(inp)
rnn = SimpleRNN(rnn_units,return_sequences=True,return_state=True,name='simple_rnn')
output_sequences, last_state = rnn(x)
out = Dense(1,activation='sigmoid',name='out')(last_state)
model = Model(inputs=inp,outputs=out)
model.compile(optimizer='adam',loss='Binary_crossEntropy',metrics=['accuracy'])
model.summary()
```
