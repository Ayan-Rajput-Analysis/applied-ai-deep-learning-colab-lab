```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from google.colab import drive
file_path = ('/content/drive/MyDrive/Project_Cricket/IPL.csv')
df = pd.read_csv(file_path)
df.head()

```

```python
df.isnull().sum()
```

```python
match_winners=df['match_winner'].value_counts()
sns.barplot(y=match_winners.index,x=match_winners.values,palette='rocket')
plt.title("Most Matched Win By Team")
```

```python
sns.countplot(x=df['toss_decision'],palette='rainbow')
plt.title('Toss Decision Trend')
```

```python
new=sns.countplot(x=df['won_by'],palette='pastel')
new
```

```python
player_of_match=df['player_of_the_match'].value_counts().head(5)
sns.barplot(x=player_of_match.values,y=player_of_match.index,palette='mako')
```

```python
top2 = df.groupby('top_scorer')['highscore'].sum().sort_values(ascending=False).head(2)
plt.figure()
top2.plot(kind='bar',color=['yellow', 'orange'])
plt.ylabel("Total Highscore")
plt.title("Top 2 Scorers")
plt.show()
```
